## Objective

The competition evaluates submissions using Mean Average Precision at 3 (MAP@3).

Before building retrieval and ranking systems, we must understand exactly how this metric rewards good predictions.

```python
def average_precision_at_3(actual, predicted):
    """
    Compute Average Precision at 3 (AP@3)
    for a single prediction.
    """

    for rank, prediction in enumerate(predicted[:3], start=1):

        if prediction == actual:
            return 1 / rank

    return 0.0
```

```python
actual = "C"

predictions = ["C","A","B"]

average_precision_at_3(actual,predictions)
```

```python
actual = "C"

predictions = ["A","C","E"]

average_precision_at_3(actual,predictions)
```

```python
actual="C"

predictions=["A","E","C"]

average_precision_at_3(actual,predictions)
```

```python
actual="C"

predictions=["A","B","D"]

average_precision_at_3(actual,predictions)
```

```python
def map_at_3(actual_labels, predicted_labels):

    scores = []

    for actual, prediction in zip(actual_labels, predicted_labels):

        scores.append(
            average_precision_at_3(actual, prediction)
        )

    return sum(scores)/len(scores)
```

```python
actual = [
    "A",
    "C",
    "B"
]

predictions = [

    ["A","C","B"],

    ["B","C","A"],

    ["D","E","B"]

]

map_at_3(actual,predictions)
```
