# Sentence Embeddings

## Objective

In this chapter we convert every document in our Knowledge Base into a dense semantic vector.

These vectors will later be indexed using FAISS for efficient semantic retrieval.

An embedding is a numerical representation of text.

Instead of comparing words directly,
we compare vectors.

This allows the system to retrieve documents
based on semantic meaning rather than exact keyword matching.
